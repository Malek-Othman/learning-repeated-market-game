## **Learning in Market Game**


```python
# Define initial values
Pi0 = 5
Pj0 = 5
Qis_0 = 100
Qjs_0 = 100
Qid_0 = 100
Qjd_0 = 100
TC0 = 1000
I0 = 1000

# Consumer reaction
S_p = [0.20, 0.10, 0.0, -0.10, -0.20]
S_c = [0.20, 0.10, 0.0, -0.10, -0.20]

max_iterations = 3  # Define the number of equilibriums to test

for t in range(1, max_iterations + 1):
    # Calculate Pit and Pjt for the current time step
    Pit = Pi0 * (1 + max(S_p))
    Pjt = Pj0 * 1.0
    Qis_t = Qis_0
    Qjs_t = Qjs_0
    # Assuming other variables are calculated within each iteration

    # Check if conditions are met
    if Pit == Pi0 * (1 + max(S_p)) and Pjt == Pj0 * 1.0:  # producer increases the price in market i with 20%
        expected_profit_t = (Qis_0 * Pit + Qjs_0 * Pjt) - TC0 * 1.05  # cost increase with 5%
        print("Eq_i_", t, "=", (-Qis_0, Pit, Pit, Qis_0), "Eq_j_", t, "=", (-Qjs_0, Pjt, Pjt, Qjs_0), "Expected profit:", expected_profit_t)

    # Calculate quantities for each market based on price changes
    if Pit == Pi0 and Pjt == Pj0:
        Qid_t = Qid_0
        Qjd_t = Qjd_0
    elif Pit != Pi0 and Pjt == Pj0:
        Qid_t = Qid_0 * (1 + max(S_c))
        Qjd_t = (I0 - Pit * Qid_t) / Pjt
    elif Pit == Pi0 and Pjt != Pj0:
        Qid_t = (I0 - Pjt * Qjd_0) / Pit
        Qjd_t = Qjd_0 * (1 + max(S_c))

    # Calculate expected income sacrificed and real profit
    Expected_It = Pit * Qid_t + Pjt * Qjd_t
    Real_profit_t = (Qid_t * Pit + Qjd_t * Pjt) - TC0 * 1.05
    print("Eq_i_", t, "=", (-Qid_t, Pit, Pit, Qid_t), "Eq_j_", t, "=", (-Qjd_t, Pjt, Pjt, Qjd_t), "Expected income:", I0 - Expected_It, "Real Profit:", Real_profit_t)

    # Check equilibrium conditions for each market
    if Qid_t > Qis_t:
        print("Market i is clear with Eq_i_", t, "=", (-Qis_t, Pit, Pit, Qis_t))
        Qid_t_plus_1 = Qid_t
    elif Qid_t < Qis_t:
        print("Market i is surplus with Eq_i_", t, "=", (-Qid_t, Pit, Pit, Qid_t), "Surplus =", Qis_t - Qid_t)
        Qid_t_plus_1 = Qid_t
    else:  # Qid_t == Qis_0
        print("Market i is at stable equilibrium with Eq_i_", t, "=", (-Qis_t, Pit, Pit, Qis_t))
        Qid_t_plus_1 = Qid_t

    if Qjd_t > Qjs_t:
        print("Market j is clear with Eq_j_", t, "=", (-Qjs_t, Pjt, Pjt, Qjs_t))
        Qjd_t_plus_1 = Qjd_t
    elif Qjd_t < Qjs_t:
        print("Market j is surplus with Eq_j_", t, "=", (-Qjd_t, Pjt, Pjt, Qjd_t), "Surplus =", Qjs_t - Qjd_t)
        Qjd_t_plus_1 = Qjd_t
    else:  # Qjd_t == Qjs_0
        print("Market j is at stable equilibrium with Eq_j_", t, "=", (-Qjs_t, Pjt, Pjt, Qjs_t))
        Qjd_t_plus_1 = Qjd_t
    print("Next Quantity to Produce is:", Qid_t_plus_1, Qjd_t_plus_1)
    # Increment t for next iteration
    t += 1 and Qis_t == Qid_t_plus_1 and Qjs_t == Qjd_t_plus_1
```

```python
# Define initial values
Pi0 = 5
Pj0= 5
Qis_0 = 100
Qjs_0 = 100
Qid_0 = 100
Qjd_0 = 100
TC0 = 1000
I0 = 1000

# Consumer reaction
S_p = [0.20, 0.10, 0.0, -0.10, -0.20]
S_c = [0.20, 0.10, 0.0, -0.10, -0.20]

max_iterations = 5  # Define the number of equilibriums to test

for t in range(1, max_iterations + 1):
    # Calculate Pit and Pjt for the current time step
    Pit = round(Pit * (1 + max(S_p)), 2)
    Pjt = round(Pjt * 1.0, 2)

    # Assuming other variables are calculated within each iteration

    # Check if conditions are met
    if Pit == round(Pit-1 * (1 + max(S_p)), 2) and Pjt == round(Pjt-1 * 1.0, 2) * 1.0:  # producer increases the price in market i with 20%
        expected_profit_t = (Qis_t-1 * Pit + Qjs_t-1 * Pjt) - TC0 * 1.05  # cost increase with 5%
        print("Eq_i_", t, "=", (-Qis_t-1, Pit, Pit, Qis_t-1), "Eq_j_", t, "=", (-Qjs_t-1, Pjt, Pjt, Qjs_t-1), "Expected profit:", expected_profit_t)

    # Calculate quantities for each market based on price changes
    if Pit == Pit-1 and Pjt == Pjt-1:
        Qid_t = Qid_t-1
        Qjd_t = Qjd_t-1
    elif Pit != Pit-1 and Pjt == Pjt-1:
        Qid_t = Qid_t-1 * (1 + max(S_c))
        Qjd_t = (I0 - Pit * Qid_t) / Pjt
    elif Pit == Pit-1 and Pjt != Pjt-1:
        Qid_t = (I0 - Pjt * Qjd_t-1) / Pit
        Qjd_t = Qjd_t-1 * (1 + max(S_c))

    # Calculate expected income sacrificed and real profit
    Expected_It = Pit * Qid_t + Pjt * Qjd_t
    Real_profit_t = (Qid_t * Pit + Qjd_t * Pjt) - TC0 * 1.05
    print("Eq_i_", t, "=", (-Qid_t, Pit, Pit, Qid_t), "Eq_j_", t, "=", (-Qjd_t, Pjt, Pjt, Qjd_t), "Expected income:", I0 - Expected_It, "Real Profit:", Real_profit_t)

    # Check equilibrium conditions for each market
    if Qid_t > Qis_t-1:
        print("Market i is clear with Eq_i_", t, "=", (-Qis_t-1, Pit, Pit, Qis_t-1))
        Qid_t_plus_1 = Qid_t
    elif Qid_t < Qis_t-1:
        print("Market i is surplus with Eq_i_", t, "=", (-Qid_t, Pit, Pit, Qid_t), "Surplus =", Qis_t-1 - Qid_t)
        Qid_t_plus_1 = Qid_t
    else:  # Qid_t == Qis_0
        print("Market i is at stable equilibrium with Eq_i_", t, "=", (-Qis_t-1, Pit, Pit, Qis_t-1))
        Qid_t_plus_1 = Qid_t

    if Qjd_t > Qjs_t-1:
        print("Market j is clear with Eq_j_", t, "=", (-Qjs_t-1, Pjt, Pjt, Qjs_t-1))
        Qjd_t_plus_1 = Qjd_t
    elif Qjd_t < Qjs_t-1:
        print("Market j is surplus with Eq_j_", t, "=", (-Qjd_t, Pjt, Pjt, Qjd_t), "Surplus =", Qjs_t-1 - Qjd_t)
        Qjd_t_plus_1 = Qjd_t
    else:  # Qjd_t == Qjs_0
        print("Market j is at stable equilibrium with Eq_j_", t, "=", (-Qjs_t-1, Pjt, Pjt, Qjs_t-1))
        Qjd_t_plus_1 = Qjd_t
        print("Next Quantity to Produce is:", "Qid_", t, "=", round(Qid_t_plus_1, 2), "Qjd_", t, "=", round(Qjd_t_plus_1, 2))

    # Increment t for next iteration
    t += 1
```
