```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import pandas as pd
import numpy as np

# ─────────────────────────────────────────────────────────────
# INITIAL STATE  S_0
# ─────────────────────────────────────────────────────────────
Qm_0 = 100
Pm_0 = 5

Qf_0 = 100
Pf_0 = 5

num_periods = 100
Cost = 1050

percentage_changes = [0.20, 0.10, 0, -0.10, -0.20]

# ─────────────────────────────────────────────────────────────
# STABILIZATION PARAMETERS
# ─────────────────────────────────────────────────────────────
MAX_Q = 300
MIN_Q = 10

MAX_P = 10
MIN_P = 1

DAMPING = 0.35

# Safe plotting ranges
quantity_range = (-MAX_Q - 20, MAX_Q + 20)
price_range = (0, MAX_P + 2)

# ─────────────────────────────────────────────────────────────
# GENERATE ALL STRATEGY POINTS
# ─────────────────────────────────────────────────────────────
def generate_all_points(Qm, Pm, Qf, Pf):

    all_points_C = []
    all_points_P = []

    for q_strategy in percentage_changes:
        for p_strategy in percentage_changes:

            Qm_new = round(Qm * (1 + q_strategy), 3)
            Pm_new = round(Pm * (1 + p_strategy), 3)

            all_points_C.append((Qm_new, Pm_new))
            all_points_P.append((-Qm_new, Pm_new))

    all_points_C_fish = []
    all_points_P_fish = []

    for q_strategy in percentage_changes:
        for p_strategy in percentage_changes:

            Qf_new = round(Qf * (1 + q_strategy), 3)
            Pf_new = round(Pf * (1 + p_strategy), 3)

            all_points_C_fish.append((Qf_new, Pf_new))
            all_points_P_fish.append((-Qf_new, Pf_new))

    return (
        all_points_C,
        all_points_P,
        all_points_C_fish,
        all_points_P_fish
    )

# ─────────────────────────────────────────────────────────────
# PRODUCER OPTIMIZATION
# ─────────────────────────────────────────────────────────────
def find_optimal_fish(Qm, Pm, Qf, Pf, cost=Cost):

    best_rev = -float("inf")
    best_coord = (Qf, Pf)

    for q_strategy in percentage_changes:
        for p_strategy in percentage_changes:

            Qf_new = round(Qf * (1 + q_strategy), 3)
            Pf_new = round(Pf * (1 + p_strategy), 3)

            revenue = Qm * Pm + Qf_new * Pf_new

            if revenue >= cost and revenue > best_rev:
                best_rev = revenue
                best_coord = (Qf_new, Pf_new)

    return best_coord

# ─────────────────────────────────────────────────────────────
# CONSUMER OPTIMIZATION
# ─────────────────────────────────────────────────────────────
def find_optimal_meat(Qm, Pm):

    best_util = -float("inf")
    best_coord = (Qm, Pm)

    for q_strategy in percentage_changes:
        for p_strategy in percentage_changes:

            Qm_new = round(Qm * (1 + q_strategy), 3)
            Pm_new = round(Pm * (1 + p_strategy), 3)

            if Pm_new <= 0:
                continue

            utility = Qm_new / Pm_new

            if utility > best_util:
                best_util = utility
                best_coord = (Qm_new, Pm_new)

    return best_coord

# ─────────────────────────────────────────────────────────────
# RUN SIMULATION
# ─────────────────────────────────────────────────────────────
states = []

Qm, Pm = Qm_0, Pm_0
Qf, Pf = Qf_0, Pf_0

for t in range(num_periods):

    all_C, all_P, all_C_fish, all_P_fish = generate_all_points(
        Qm, Pm, Qf, Pf
    )

    opt_fish = find_optimal_fish(Qm, Pm, Qf, Pf)
    opt_meat = find_optimal_meat(Qm, Pm)

    states.append({
        "period": t + 1,
        "Qm": Qm,
        "Pm": Pm,
        "Qf": Qf,
        "Pf": Pf,
        "opt_meat": opt_meat,
        "opt_fish": opt_fish,
        "all_C": all_C,
        "all_P": all_P,
        "all_C_fish": all_C_fish,
        "all_P_fish": all_P_fish,
    })

    # ─────────────────────────────────────────
    # DAMPED UPDATE
    # ─────────────────────────────────────────

    Qm = Qm + DAMPING * (opt_meat[0] - Qm)
    Pm = Pm + DAMPING * (opt_meat[1] - Pm)

    Qf = Qf + DAMPING * (opt_fish[0] - Qf)
    Pf = Pf + DAMPING * (opt_fish[1] - Pf)

    # Bound values

    Qm = np.clip(Qm, MIN_Q, MAX_Q)
    Pm = np.clip(Pm, MIN_P, MAX_P)

    Qf = np.clip(Qf, MIN_Q, MAX_Q)
    Pf = np.clip(Pf, MIN_P, MAX_P)

# ─────────────────────────────────────────────────────────────
# DRAW SINGLE PERIOD
# ─────────────────────────────────────────────────────────────
def draw_period(t_idx, axes):

    s = states[t_idx]

    all_points_C = s["all_C"]
    all_points_P = s["all_P"]

    all_points_C_fish = s["all_C_fish"]
    all_points_P_fish = s["all_P_fish"]

    optimal_fish_coord = s["opt_fish"]
    optimal_meat_coord = s["opt_meat"]

    period_num = s["period"]

    for i, ax in enumerate(axes):

        ax.clear()

        ax.set_xlim(quantity_range)
        ax.set_ylim(price_range)

        ax.grid(True, alpha=0.3)

        ax.set_xlabel("Quantity")
        ax.set_ylabel("Price")

        # Initial equilibrium

        ax.scatter(
            [-Qm_0, -Qf_0],
            [Pm_0, Pf_0],
            color='red',
            s=80,
            zorder=5,
            label='Initial Equilibrium'
        )

        ax.scatter(
            [Qm_0, Qf_0],
            [Pm_0, Pf_0],
            color='red',
            s=80,
            zorder=5
        )

        # Blue strategy clouds

        if i == 0:

            for point in all_points_P:
                ax.scatter(
                    point[0],
                    point[1],
                    color='blue',
                    s=15,
                    alpha=0.4
                )

            for point in all_points_C:
                ax.scatter(
                    point[0],
                    point[1],
                    color='blue',
                    s=15,
                    alpha=0.4
                )

            ax.set_title(
                f'Market x1 (meat) — Period {period_num}'
            )

        else:

            for point in all_points_P_fish:
                ax.scatter(
                    point[0],
                    point[1],
                    color='blue',
                    s=15,
                    alpha=0.4
                )

            for point in all_points_C_fish:
                ax.scatter(
                    point[0],
                    point[1],
                    color='blue',
                    s=15,
                    alpha=0.4
                )

            ax.set_title(
                f'Market x2 (fish) — Period {period_num}'
            )

        # New equilibrium

        if i == 0:

            ax.scatter(
                -optimal_meat_coord[0],
                optimal_meat_coord[1],
                color='green',
                s=120,
                zorder=6,
                label='Optimal equilibrium'
            )

            ax.text(
                -optimal_meat_coord[0],
                optimal_meat_coord[1],
                str(period_num),
                fontsize=9
            )

        else:

            ax.scatter(
                -optimal_fish_coord[0],
                optimal_fish_coord[1],
                color='green',
                s=120,
                zorder=6
            )

            ax.text(
                -optimal_fish_coord[0],
                optimal_fish_coord[1],
                str(period_num),
                fontsize=9
            )

        ax.legend(fontsize=8)

# ─────────────────────────────────────────────────────────────
# ANIMATION
# ─────────────────────────────────────────────────────────────
def run_animation():

    fig, axes = plt.subplots(
        ncols=2,
        figsize=(12, 5),
        constrained_layout=True
    )

    fig.suptitle(
        'Market equilibrium — 100-period simulation',
        fontsize=13
    )

    def update(frame):
        draw_period(frame, axes)

    anim = animation.FuncAnimation(
        fig,
        update,
        frames=num_periods,
        interval=400,
        repeat=True,
        cache_frame_data=False
    )

    return anim

# ─────────────────────────────────────────────────────────────
# STATIC SNAPSHOTS
# ─────────────────────────────────────────────────────────────
def run_static_grid(step=20):

    snapshots = list(range(0, num_periods, step))

    n_cols = len(snapshots)

    fig, big_axes = plt.subplots(
        nrows=2,
        ncols=n_cols,
        figsize=(4 * n_cols, 7),
        constrained_layout=True
    )

    fig.suptitle(
        'Market equilibrium snapshots',
        fontsize=13
    )

    for col_idx, t_idx in enumerate(snapshots):

        ax_pair = [
            big_axes[0][col_idx],
            big_axes[1][col_idx]
        ]

        draw_period(t_idx, ax_pair)

# ─────────────────────────────────────────────────────────────
# PATH PLOT
# ─────────────────────────────────────────────────────────────
def run_path_plot():

    fig, axes = plt.subplots(
        ncols=2,
        figsize=(12, 5),
        constrained_layout=True
    )

    fig.suptitle(
        'Equilibrium path over 100 periods',
        fontsize=13
    )

    meat_Q = [s["opt_meat"][0] for s in states]
    meat_P = [s["opt_meat"][1] for s in states]

    fish_Q = [s["opt_fish"][0] for s in states]
    fish_P = [s["opt_fish"][1] for s in states]

    periods = [s["period"] for s in states]

    plots = [
        (
            axes[0],
            meat_Q,
            meat_P,
            'Market x1 (meat)'
        ),
        (
            axes[1],
            fish_Q,
            fish_P,
            'Market x2 (fish)'
        )
    ]

    for ax, Q_path, P_path, title in plots:

        scatter = ax.scatter(
            Q_path,
            P_path,
            c=periods,
            cmap='viridis',
            s=50,
            zorder=4
        )

        ax.plot(
            Q_path,
            P_path,
            color='gray',
            alpha=0.5
        )

        ax.grid(True, alpha=0.3)

        ax.set_title(title)
        ax.set_xlabel('Quantity')
        ax.set_ylabel('Price')

        plt.colorbar(
            scatter,
            ax=ax,
            label='Period'
        )

# ─────────────────────────────────────────────────────────────
# SUMMARY TABLE
# ─────────────────────────────────────────────────────────────
def print_summary(every=10):

    rows = []

    for s in states[::every]:

        rows.append({
            "Period": s["period"],
            "Qm": round(s["Qm"], 2),
            "Pm": round(s["Pm"], 2),
            "Qf": round(s["Qf"], 2),
            "Pf": round(s["Pf"], 2),
            "Revenue": round(
                s["Qm"] * s["Pm"] + s["Qf"] * s["Pf"],
                2
            )
        })

    df_summary = pd.DataFrame(rows)

    print(df_summary.to_string(index=False))

# ─────────────────────────────────────────────────────────────
# MAIN
# ─────────────────────────────────────────────────────────────
if __name__ == "__main__":

    print_summary(every=10)

    anim = run_animation()

    run_static_grid(step=20)

    run_path_plot()

    plt.show()
```

```python
import pycid

# Création d'un Multi-Agent Influence Diagram (MAID) pour le marché du travail
job_market = pycid.MACID(
    [
        ("T", "D1"),  # Relation entre T (travail) et D1 (décision 1)
        ("T", "U1"),  # Relation entre T (travail) et U1 (utilité 1)
        ("T", "U2"),  # Relation entre T (travail) et U2 (utilité 2)
        ("D1", "D2"), # Relation entre D1 (décision 1) et D2 (décision 2)
        ("D1", "U1"), # Relation entre D1 (décision 1) et U1 (utilité 1)
        ("D2", "U1"), # Relation entre D2 (décision 2) et U1 (utilité 1)
        ("D2", "U2"), # Relation entre D2 (décision 2) et U2 (utilité 2)
    ],
    agent_decisions={
        1: ["D1"], # Décision de l'agent 1 : D1
        2: ["D2"], # Décision de l'agent 2 : D2
    },
    agent_utilities={
        1: ["U1"], # Utilité de l'agent 1 : U1
        2: ["U2"], # Utilité de l'agent 2 : U2
    },
)

# Dessin du MAID représentant le marché du travail
job_market.draw()

# Définition de la probabilité
prob = 1/2

# Ajout des Conditional Probability Distributions (CPDs)
job_market.add_cpds(
    T=pycid.bernoulli(prob), # Distribution de probabilité pour T (travail)
    D1=[0, 1],               # Distribution de probabilité pour D1 (décision 1)
    D2=[0, 1],               # Distribution de probabilité pour D2 (décision 2)
    U1=lambda D1, D2, T: 5 * D2 - T * D1 - 2 * D1 * (1 - T), # Utilité de l'agent 1
    U2=lambda D2, T: 3 * T * D2 - 2 * (1 - T) * D2 - (1 - D2) * T, # Utilité de l'agent 2
)
```

```python
import numpy as np
import random

# ── Environment ────────────────────────────────────────────────────────────────

class ExtensiveGameEnvironment:
    """Simple two-player zero-sum game (e.g. resource competition)."""

    def __init__(self, num_rounds=10):
        self.num_rounds = num_rounds
        self.num_actions = 3          # e.g. cooperate / neutral / defect
        self.num_agents = 2
        self.round = 0
        self.state = None

    def reset(self):
        self.round = 0
        self.state = np.zeros(self.num_agents, dtype=int)
        return self.state.copy()

    def step(self, actions):
        # Reward logic: reward agents who defect vs cooperators, penalise mutual defection
        rewards = [0.0] * self.num_agents
        for i in range(self.num_agents):
            for j in range(self.num_agents):
                if i != j:
                    rewards[i] += actions[i] - actions[j]   # higher action = more aggressive
        rewards = [r / max(1, self.num_agents - 1) for r in rewards]

        self.state = np.array(actions, dtype=int)
        self.round += 1
        done = self.round >= self.num_rounds
        return self.state.copy(), rewards, done


# ── Agent ──────────────────────────────────────────────────────────────────────

class Agent:
    """Q-learning agent with an epsilon-greedy policy."""

    def __init__(self, num_actions=3, state_size=2, lr=0.1, gamma=0.95,
                 epsilon=1.0, epsilon_decay=0.995, epsilon_min=0.05):
        self.num_actions  = num_actions
        self.lr           = lr
        self.gamma        = gamma
        self.epsilon      = epsilon
        self.epsilon_decay = epsilon_decay
        self.epsilon_min  = epsilon_min

        # Q-table keyed by (state tuple) → action values
        self.q_table = {}

    def _state_key(self, state):
        return tuple(state)

    def _get_q(self, state):
        key = self._state_key(state)
        if key not in self.q_table:
            self.q_table[key] = np.zeros(self.num_actions)
        return self.q_table[key]

    def choose_action(self, state):
        if random.random() < self.epsilon:
            return random.randrange(self.num_actions)       # explore
        return int(np.argmax(self._get_q(state)))           # exploit

    def update_policy(self, state, action, reward, next_state):
        q_values      = self._get_q(state)
        q_next        = self._get_q(next_state)
        td_target     = reward + self.gamma * np.max(q_next)
        q_values[action] += self.lr * (td_target - q_values[action])

        # Decay exploration rate
        self.epsilon = max(self.epsilon_min, self.epsilon * self.epsilon_decay)


# ── Training ───────────────────────────────────────────────────────────────────

num_agents   = 2
num_episodes = 1000

env    = ExtensiveGameEnvironment()
agents = [Agent() for _ in range(num_agents)]

episode_rewards = [[] for _ in range(num_agents)]

for episode in range(num_episodes):
    state = env.reset()
    done  = False
    ep_rewards = [0.0] * num_agents

    while not done:
        actions                  = [agent.choose_action(state) for agent in agents]
        next_state, rewards, done = env.step(actions)

        for i, agent in enumerate(agents):
            agent.update_policy(state, actions[i], rewards[i], next_state)
            ep_rewards[i] += rewards[i]

        state = next_state

    for i in range(num_agents):
        episode_rewards[i].append(ep_rewards[i])

    if (episode + 1) % 100 == 0:
        avg = [np.mean(episode_rewards[i][-100:]) for i in range(num_agents)]
        print(f"Episode {episode+1:>4} | Avg rewards (last 100): "
              + " | ".join(f"Agent {i}: {avg[i]:+.3f}" for i in range(num_agents)))


# ── Evaluation ─────────────────────────────────────────────────────────────────

print("\n── Evaluation (50 episodes, no exploration) ──")
eval_rewards = [0.0] * num_agents

for _ in range(50):
    state = env.reset()
    done  = False
    while not done:
        # Greedy actions only (epsilon → 0)
        actions = [int(np.argmax(agent._get_q(state))) for agent in agents]
        next_state, rewards, done = env.step(actions)
        for i in range(num_agents):
            eval_rewards[i] += rewards[i]
        state = next_state

for i in range(num_agents):
    print(f"Agent {i} total eval reward: {eval_rewards[i]:.3f}")
```
